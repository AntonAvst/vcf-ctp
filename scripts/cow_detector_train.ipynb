# Train

```python
!yolo detect train \
  model=yolov8s.pt \
  data="$DATA_YAML" \
  imgsz=1280 \
  epochs=100 \
  batch=16 \
  workers=2
```

# Validate

```python
!yolo detect val model=runs/detect/train/weights/best.pt data="$DATA_YAML" imgsz=1280
!yolo detect predict model=runs/detect/train/weights/best.pt source="$DATA_YAML".replace("data.yaml","valid/images") save=True
```
